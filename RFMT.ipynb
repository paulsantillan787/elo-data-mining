```python
def assign_score(val, bins, reverse=False):
    if reverse:  # Recency, menor es mejor
        if val <= bins[0]: return 4
        elif val <= bins[1]: return 3
        elif val <= bins[2]: return 2
        else: return 1
    else:  # Frequency, Monetary, Tenure: mayor es mejor
        if val <= bins[0]: return 1
        elif val <= bins[1]: return 2
        elif val <= bins[2]: return 3
        else: return 4
```

```python
df_pandas['R_score'] = df_pandas['recency_days'].apply(lambda x: assign_score(x, r_bins, reverse=True))
df_pandas['F_score'] = df_pandas['frequency'].apply(lambda x: assign_score(x, f_bins))
df_pandas['M_score'] = df_pandas['monetary_winsor'].apply(lambda x: assign_score(x, m_bins))
df_pandas['T_score'] = df_pandas['tenure_days'].apply(lambda x: assign_score(x, t_bins))
```

```python
import pandas as pd

# Supón que tu DataFrame se llama df_rfmt y tiene las columnas 'RFMT' y 'count'
def describir_rfmt(rfmt):
    # Cada dígito del código
    r, f, m, t = [int(x) for x in str(rfmt)]
    texto = []
    # Recency
    if r == 1:
        texto.append("Cliente muy reciente")
    elif r == 2:
        texto.append("Cliente reciente")
    elif r == 3:
        texto.append("Cliente algo inactivo")
    else:
        texto.append("Cliente inactivo")
    # Frequency
    if f == 4:
        texto.append("compra muy frecuentemente")
    elif f == 3:
        texto.append("compra frecuentemente")
    elif f == 2:
        texto.append("compra pocas veces")
    else:
        texto.append("casi no compra")
    # Monetary
    if m == 4:
        texto.append("y gasta mucho")
    elif m == 3:
        texto.append("y gasta por encima del promedio")
    elif m == 2:
        texto.append("y gasta poco")
    else:
        texto.append("y gasta muy poco")
    # Tenure
    if t == 4:
        texto.append("con larga relación")
    elif t == 3:
        texto.append("con relación estable")
    elif t == 2:
        texto.append("con relación reciente")
    else:
        texto.append("nuevo en la empresa")
    return ", ".join(texto)

df_rfmt['Descripción'] = df_rfmt['RFMT'].apply(describir_rfmt)

# Opcional: ordenar por cantidad descendente
df_rfmt = df_rfmt.sort_values(by='count', ascending=False)

# Visualizar
print(df_rfmt[['RFMT', 'count', 'Descripción']])
```

```python
import pandas as pd
import numpy as np

# Función para calcular el R² explicado por la segmentación
def rfmt_explicabilidad(df, cortes, mostrar_top=10):
    # Aplica los cortes
    df = df.copy()
    df['R'] = pd.cut(df['recency_days'], bins=[-np.inf]+cortes['recency']+[np.inf], labels=[4,3,2,1], right=True).astype(float).fillna(1).astype(int)
    df['F'] = pd.cut(df['frequency'], bins=[-np.inf]+cortes['frequency']+[np.inf], labels=[1,2,3,4], right=True).astype(float).fillna(1).astype(int)
    df['M'] = pd.cut(df['monetary_winsor'], bins=[-np.inf]+cortes['monetary']+[np.inf], labels=[1,2,3,4], right=True).astype(float).fillna(1).astype(int)
    df['T'] = pd.cut(df['tenure_days'], bins=[-np.inf]+cortes['tenure']+[np.inf], labels=[1,2,3,4], right=True).astype(float).fillna(1).astype(int)
    df['RFMT'] = df['R'].astype(str) + df['F'].astype(str) + df['M'].astype(str) + df['T'].astype(str)
    
    # Media global y varianza
    mean_target_global = df['target'].mean()
    ss_total = ((df['target'] - mean_target_global) ** 2).sum()
    
    # Media por grupo
    grouped = df.groupby('RFMT').agg(
        group_count=('target', 'size'),
        group_mean=('target', 'mean')
    ).reset_index()
    ss_between = ((grouped['group_mean'] - mean_target_global) ** 2 * grouped['group_count']).sum()
    r2 = ss_between / ss_total

    # Muestra resultados
    print(f"Cortes: {cortes}")
    print(f"Varianza explicada (R²): {r2:.2%}")
    print("Top grupos por tamaño:")
    print(grouped.sort_values("group_count", ascending=False).head(mostrar_top))
    print()
    return r2, grouped

# EJEMPLO DE USO:
# Define aquí tus cortes (puedes ir cambiando y volver a correr)
cortes = {
    'recency':   [7, 30, 90],       # 4 grupos: <=7, 8-30, 31-90, >90
    'frequency': [20, 60, 120],     # 1-20, 21-60, 61-120, >120
    'monetary':  [-1.35, -1.21, -0.82], # <=-1.35, -1.35 a -1.21, -1.21 a -0.82, >-0.82
    'tenure':    [180, 285, 415],   # <=180, 181-285, 286-415, >415
}

# Supón que tu DataFrame se llama df_pandas y tiene las columnas correctas
r2, grupos = rfmt_explicabilidad(df_pandas, cortes)

# Cambia los valores en 'cortes' para experimentar y observa el R² y los grupos resultantes.
```

```python
import pandas as pd

# Asumiendo que tu DataFrame se llama df y tiene las columnas 'target' y 'super_segment'

# 1. Media global y suma de cuadrados total
mean_target = df_pandas['target'].mean()
ss_total = ((df_pandas['target'] - mean_target) ** 2).sum()

# 2. Media y tamaño por super_segment
grouped = df_pandas.groupby('super_segment').agg(
    group_count=('target', 'size'),
    group_mean=('target', 'mean')
).reset_index()

# 3. Suma de cuadrados explicada por los grupos
ss_between = ((grouped['group_mean'] - mean_target) ** 2 * grouped['group_count']).sum()

# 4. R² explicado
r2 = ss_between / ss_total
print(f"Varianza explicada solo por los súper-segmentos: {r2:.2%}")
```

La segmentación RFMT detallada explica el 5.6% de la varianza de la variable target. Sin embargo, al agrupar estos perfiles en súper-segmentos accionables, el porcentaje de varianza explicada se reduce a 0.34%. Esto confirma que existe una pérdida de capacidad explicativa al simplificar la segmentación, pero se gana en claridad y utilidad para la toma de decisiones estratégicas.

```python
def super_segment(row):
    # Desglosa RFMT (asume que RFMT es string de 4 dígitos)
    r, f, m, t = map(int, list(str(row['RFMT']).zfill(4)))
    # Lógica de súper-segmentos (ajusta reglas según tu negocio)
    if r == 4 and f == 4 and m == 4 and t == 4:
        return 'Premium'
    elif r == 4 and t <= 2 and f >= 3 and m >= 3:
        return 'Nuevos Prometedores'
    elif r >= 3 and f >= 3 and t >= 3 and m >= 2 and m <= 3:
        return 'Leales y constantes'
    elif r >= 3 and f >= 2 and m >= 3 and t >= 3:
        return 'Alto potencial'
    elif t <= 2 and r >= 3 and f >= 2 and m >= 2:
        return 'Crecimiento'
    elif r in [2,3] and f == 2 and m == 2 and t in [2,3]:
        return 'Oportunidad'
    elif r <= 2 and t >= 3:
        return 'Dormidos'
    elif m <= 2 and f in [2,3] and t >= 2:
        return 'Sensibles al precio'
    elif f == 1 and m <= 2 and t <= 2 and r in [2,3]:
        return 'Esporádicos'
    elif r == 1:
        return 'Riesgo/Bajo valor'
    else:
        return 'Medio'  # Resto de clientes

# Aplica la función a tu DataFrame
df_pandas['super_segment'] = df_pandas.apply(super_segment, axis=1)
```
